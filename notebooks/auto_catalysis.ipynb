# Auto catalysis

in some reactions a chemical entity can produce more of itself by adding something else and initiating the reaction, that's auto catalysis


```python
model = Autocatalysis()
```

```python
k_a=1 # forward direction rate constant
k_b=1 # reverse direction rate constant
a = 1 # concerntration of reactant a (which aids the production of X (our auto catalyst, chemists can correct on the wording here lol))
```

```python
dxdt = lambda x: model.rhs(x, k_a=k_a, k_b=k_b, a=a)
```

```python
# dxdt = lambda x: x*r
```

```python
tn=1
dt=0.01
x0 = [0,1, 5, 10,15, 20,25]
```

```python
dxdt = lambda x: model.rhs(x, k_a=k_a, k_b=k_b, a=a)
ax = plot_solution(dxdt, x0=x0, tn=tn, dt=dt, n_slope=20, slope_length=0.2, show=True, title="autocatalysis")
```

```python
# now lets see the mode behavior at different values of rate constants
```

```python
k_a=1
k_b=4
a=10
dxdt = lambda x: model.rhs(x, k_a=k_a, k_b=k_b, a=a)
print('fp=', a*(k_a / k_b))
ax = plot_solution(dxdt, x0=x0, tn=tn, dt=dt, n_slope=20, slope_length=0.2, show=True, title="autocatalysis")
```

```python
k_a=3
k_b=4
a=10
dxdt = lambda x: model.rhs(x, k_a=k_a, k_b=k_b, a=a)
print('fp=', a*(k_a / k_b))
ax = plot_solution(dxdt, x0=x0, tn=tn, dt=dt, n_slope=20, slope_length=0.2, show=True, title="autocatalysis")
```

```python
# we also know from finding the fixed points that... x* = a*(k_a / k_b)
# as i see if my reverse reaction rate is slow and my quantity of reactant a remains constant my equilibrium will
# be at a higher point (still constant a here acts a stopping force so a runaway reaction won't happen, the FP just will be higher)
# e.g lets see two ridiculous cases (high k_b and high k_a); if k_a is bigger my forward reaction is faster 
# so ill have higher concentration for equilibrium and vice versa for high k_b
```

```python
tn=1
dt=0.001 
x0 = [0,1, 5, 10,15, 20,25]
k_a=10
k_b=1
a=1
dxdt = lambda x: model.rhs(x, k_a=k_a, k_b=k_b, a=a)
ax = plot_solution(dxdt, x0=x0, tn=tn, dt=dt, n_slope=20, slope_length=0.2, show=True, title="autocatalysis\nhigh forward rate constant")
print('fp=', a*(k_a / k_b))
```
