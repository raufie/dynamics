# Auto catalysis

in some reactions a chemical entity can produce more of itself by adding something else and initiating the reaction, that's auto catalysis


```python
model = TumorGrowth()
```

```python
a=1 # intrinsic growth/rate of growth (bigger means itll reach FP faster, i.e how fast the tumor grows)
b=(1/10) # carrying capacity (k=1/b, so put in 1/K); found via analyzing FPs
```

```python
dxdt = lambda x: model.rhs(x, a=a, b=b)
```
